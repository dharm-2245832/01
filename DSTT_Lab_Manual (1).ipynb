### 8. Create a new column performance: If `salary > 10000`, set it to "High".

```python
df['Performance'] = np.where(df['Annual Salary'] > 10000, "High", "")
df.head()
```

### 9. how to identify duplicate & drop duplicates keeping only the first occurrence using pandas.
