```python
customer_reviews = ['sam was a great help to me in the store', 
                    'the cashier was very rude to me, I think her name was eleanor', 
                    'amazing work from sadeen!', 
                    'sarah was able to help me find the items i needed quickly', 
                    'lucy is such a great addition to the team', 
                    'great service from sara she found me what i wanted'
                   ]
```

only sarah's review

```python
sarah_review = []
pattern_to_find = r"sarah?"
```

```python
for string in customer_reviews:
    if(re.search(pattern_to_find, string)):
        sarah_review.append(string)
```

```python
sarah_review
```

#### review that start with the letter a

```python
a_review = []
pattern_to_find_a = r"^a"
```

```python
for string in customer_reviews:
    if(re.search(pattern_to_find_a, string)):
        a_review.append(string)
```

```python
a_review
```

#### review that end with the letter a

```python
y_review = []
pattern_to_find_y = r"y$"
```

```python
for string in customer_reviews:
    if(re.search(pattern_to_find_y, string)):
        y_review.append(string)
```

```python
y_review
```

#### review that contains the words need or wanted

```python
need_review = []
pattern_to_find_need = r"(need|want)ed"
```

```python
for string in customer_reviews:
    if(re.search(pattern_to_find_need, string)):
        need_review.append(string)
```

```python
need_review
```

```python
no_punct_review = []
pattern_to_find_no_punct = r"^w\s"
```

```python
for string in customer_reviews:
    if(re.sub(pattern_to_find_no_punct,"", string)):
        no_punct_review.append(string)
```

```python
no_punct_review
```

### Experiment 7 → Bag of Words

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
```

```python
data = [' Most shark attacks occur about 10 feet from the beach since that is where the people are',
        'the efficiency with which he paired the socks in the drawer was quite admirable',
        'carol drank the blood as if she were a vampire',
        'giving directions that the mountains are to the west only works when you can see them',
        'the sign said there was road work ahead so he decided to speed up',
        'the gruff old man sat in the back of the bait shop grumbling to himself as he scooped out a handful of worms']
```

```python
countvec = CountVectorizer()
```

```python
countvet_fit = countvec.fit_transform(data)
```

```python
bag_of_words = pd.DataFrame(countvet_fit.toarray(), columns = countvec.get_feature_names_out())
```

```python
bag_of_words
```

## Experiment 10

```python
import nltk
from nltk.stem import SnowballStemmer, RegexpStemmer
```

```python
words = ["runes", "happily", "jumping", "easily", "faster","going","walking"]
```

```python
snowball = SnowballStemmer("english")
```

```python
snowball_stems = [snowball.stem(word) for word in words]
```

```python
regexp = RegexpStemmer('ing$|ly$|s$', min=4)  
```

```python
regexp_stems = [regexp.stem(word) for word in words]
```

```python
print("Original Words:   ", words)
print("Snowball Stems:   ", snowball_stems)
print("Regexp Stems:     ", regexp_stems)
```
