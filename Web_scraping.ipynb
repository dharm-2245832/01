```python
# for all 10 pages
p_name1 = []
prices1 = []
desc1 = []

for i in range(2,12):
    url = "https://www.flipkart.com/search?q=mobile+under+10000&sid=tyy%2C4io&as=on&as-show=on&otracker=AS_QueryStore_OrganicAutoSuggest_1_12_na_na_na&otracker1=AS_QueryStore_OrganicAutoSuggest_1_12_na_na_na&as-pos=1&as-type=RECENT&suggestionId=mobile+under+10000%7CMobiles&requestId=bb951fc9-48a4-47a7-87d5-22473e516e86&as-searchtext=mobile+under+10000&page="+str(i)
    r = requests.get(url1)
    soup1 = BeautifulSoup(r.text, "lxml")
    box1 = soup1.find("div", class_="DOjaWF gdgoEp")

    names1 = box.find_all("div",class_="KzDlHZ")
    for i in names1:
        n = i.text
        p_name1.append(n)

    price1 = box.find_all("div", class_="Nx9bqj _4b5DiR")
    for i in price1:
        p = i.text
        prices1.append(p)

    dl1 = box.find_all("ul", class_="G4BRas")
    for ul in dl1:
        features1 = [li.text for li in ul.find_all("li")]
        desc1.append(features1)
```

```python
webscap1 = pd.DataFrame({"Product_name": p_name1, "Prices":prices1, "Description":desc1})
webscap1
```
